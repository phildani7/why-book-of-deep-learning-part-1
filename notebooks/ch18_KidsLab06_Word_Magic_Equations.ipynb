# 🧙‍♂️ Word Magic: The Secret Math of Words! 🎯

## Welcome, Young Explorer! 👋

Did you know that computers can do **MATH with WORDS**?

Just like: `5 − 3 + 2 = 4`

We can do: `KING − MAN + WOMAN = QUEEN`

### 🤔 The Big Idea
Every word has a secret address made of numbers. When we add and subtract words, we're really adding and subtracting their number-addresses!

````python
# 📖 Loading the word dictionary
import gensim.downloader as api

print("⏳ Downloading word dictionary... (this takes a minute)")
print("   Go get a snack! 🍪")
print()

# Using GloVe - smaller and faster!
model = api.load('glove-wiki-gigaword-100')

print("✅ Loaded! We now know", len(model), "words!")
print("🎉 Ready for word magic!")
````

---
## 👑 Step 4: The Famous King-Queen Equation!

Here's the amazing discovery:

# $$\text{KING} - \text{MAN} + \text{WOMAN} = \ ???$$

What do you think the answer will be? Let's find out!

````python
# 👑 THE FAMOUS KING-QUEEN EQUATION!

print("═"*50)
print("🧮 EQUATION: king − man + woman = ???")
print("═"*50)
print()

# Do the magic math!
result = model.most_similar(positive=['king', 'woman'], negative=['man'], topn=5)

print("🎯 TOP ANSWERS:")
print()
for i, (word, score) in enumerate(result, 1):
    stars = "⭐" * (6-i)
    print(f"   {i}. {word.upper()} {stars} (score: {score:.2f})")

print()
if result[0][0] == "queen":
    print("🎉 WOW! The computer figured out QUEEN!")
else:
    print(f"🤔 Interesting! This time the top answer is {result[0][0].upper()}.")
print()
print("💡 WHY? Because:")
print("   • KING has 'royalty' + 'male'")
print("   • Subtract 'male' (by removing MAN)")
print("   • Add 'female' (by adding WOMAN)")
print("   • You get QUEEN! 👸")
````

### The Formula:
# $$\text{WORD}_1 - \text{WORD}_2 + \text{WORD}_3 = \ ???$$

````python
# 🛠️ Our magic equation solver

def solve_equation(word1, word2, word3, emoji="🎯"):
    """
    Solves: word1 − word2 + word3 = ???
    """
    print("═"*50)
    print(f"{emoji} EQUATION: {word1} − {word2} + {word3} = ???")
    print("═"*50)

    try:
        result = model.most_similar(positive=[word1, word3], negative=[word2], topn=5)
        print("\n🎯 TOP 5 ANSWERS:")
        for i, (word, score) in enumerate(result, 1):
            medal = ["🥇", "🥈", "🥉", "4️⃣", "5️⃣"][i-1]
            print(f"   {medal} {word.upper()} (score: {score:.2f})")
    except KeyError as e:
        print(f"\n❌ Oops! The word {e} is not in our dictionary.")
        print("   Try a different word!")

    print()

print("✅ Equation solver ready! Let's try the puzzles below! 👇")
````

## 📊 ALL 10 EQUATIONS - Summary

Here are our **guesses** next to what the word dictionary **really** says. Run the cell below and look closely:
some equations come out perfectly, some find the answer in 2nd or 3rd place, and some miss it completely.
That is not a bug in your code! Word vectors learn from how words are *used* in real text, and real text does
not always follow our neat little rules. (Here is a clue: "day" and "days" are used in almost the same
sentences, so they live very close together.)

````python
# 📊 Our guesses vs what the dictionary really says
equations = [("dog", "puppy", "kitten", "cat"),     ("france", "paris", "tokyo", "japan"),
             ("smaller", "small", "big", "bigger"), ("walked", "walk", "run", "ran"),
             ("hindi", "india", "france", "french"), ("father", "man", "woman", "mother"),
             ("nest", "bird", "human", "house"),    ("yellow", "banana", "apple", "red"),
             ("cold", "hot", "day", "night"),       ("school", "teacher", "doctor", "hospital")]

print(f"{'#':>2}  {'equation':32} {'our guess':10} {'top answer':12} where our guess came")
print("─" * 78)
hits = 0
for k, (w1, w2, w3, guess) in enumerate(equations, 1):
    top5 = [w for w, _ in model.most_similar(positive=[w1, w3], negative=[w2], topn=5)]
    place = top5.index(guess) + 1 if guess in top5 else None
    hits += place == 1
    where = {1: "🥇 1st!", 2: "🥈 2nd", 3: "🥉 3rd"}.get(place, f"#{place}" if place else "not in the top 5")
    print(f"{k:>2}  {w1 + ' − ' + w2 + ' + ' + w3:32} {guess:10} {top5[0]:12} {where}")
print(f"\n🎯 Our guess came out on top in {hits} of {len(equations)} equations.")
````

## 🔬 BONUS: Find Word Neighbors!

What words live closest to any word? Let's find out!

````python
# 🔬 Find the nearest neighbors of any word!

def find_neighbors(word, n=10):
    """Find the n closest words to the given word"""
    print(f"🏘️ The {n} closest neighbors to '{word.upper()}':")
    print()

    try:
        neighbors = model.most_similar(word, topn=n)
        for i, (neighbor, score) in enumerate(neighbors, 1):
            bar = "█" * int(score * 20)
            print(f"   {i:2}. {neighbor:15} {bar} ({score:.2f})")
    except KeyError:
        print(f"   ❌ '{word}' is not in our dictionary. Try another word!")

    print()

# Try it! Change the word below:
find_neighbors("happy")
````

## 🧮 SUPER BONUS: See the Actual Math!

Want to see the actual addition and subtraction of numbers?

Watch for a surprise: the word closest to `king − man + woman` is... **KING** itself! Taking away *man* and
adding *woman* only nudges the king vector a little, so it is still nearest to where it started.
`most_similar` (which we used above) quietly **leaves out the words you typed in**. Do the same here and QUEEN
appears.

````python
# 🧮 See the actual vector math!
import numpy as np

# Get the vectors (list of numbers) for each word
king_vector = model['king']
man_vector = model['man']
woman_vector = model['woman']

# Do the math: king - man + woman
result_vector = king_vector - man_vector + woman_vector

print("🧮 THE ACTUAL MATH:")
print()
print(f"   KING vector (first 5 numbers):   {king_vector[:5].round(2)}")
print(f" − MAN vector (first 5 numbers):    {man_vector[:5].round(2)}")
print(f" + WOMAN vector (first 5 numbers):  {woman_vector[:5].round(2)}")
print(f"   " + "─"*50)
print(f" = RESULT (first 5 numbers):        {result_vector[:5].round(2)}")
print()

# Now find what word is closest to this result
similar = model.similar_by_vector(result_vector, topn=4)
print("🎯 The words closest to this result:")
for word, score in similar:
    print(f"   → {word.upper()} (score: {score:.2f})")
print()
print("😮 KING is still closest! The arrow only moved a little.")
answer = [(w, s) for w, s in similar if w not in ("king", "man", "woman")][0]
print(f"   Leave out the three words we typed in (as most_similar does): {answer[0].upper()} ({answer[1]:.2f})")
````

````python
# 🧮 Try your own vector math! Change the words below:

def show_vector_math(word1, word2, word3):
    """Shows the actual number math for: word1 - word2 + word3"""
    try:
        v1 = model[word1]
        v2 = model[word2]
        v3 = model[word3]
        result = v1 - v2 + v3

        print(f"🧮 VECTOR MATH: {word1} − {word2} + {word3}")
        print()
        print(f"   {word1.upper():10} = {v1[:4].round(2)} ...")
        print(f" − {word2.upper():10} = {v2[:4].round(2)} ...")
        print(f" + {word3.upper():10} = {v3[:4].round(2)} ...")
        print(f"   " + "─"*40)
        print(f" = RESULT      = {result[:4].round(2)} ...")
        print()

        similar = model.similar_by_vector(result, topn=8)
        print("🎯 Closest words to result:")
        for word, score in similar[:3]:
            print(f"   → {word.upper()} ({score:.2f})")
        best = [(w, s) for w, s in similar if w not in (word1, word2, word3)][0]
        print(f"   Leaving out the words we typed in: {best[0].upper()} ({best[1]:.2f})")
    except KeyError as e:
        print(f"❌ Word {e} not found!")

# Try it:
show_vector_math("father", "man", "woman")
````
